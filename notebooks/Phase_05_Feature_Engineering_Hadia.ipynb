```python
#import required libraries
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
# Load the preprocessed dataset from Phase 4

df = pd.read_csv("Preprocessed_Reviews_File.csv")
```

```python
# Check for missing values in the Processed Text column

missing_values = df["Processed Text"].isnull().sum()

print("Missing values in Processed Text:", missing_values)
```

```python
# Display the first five records

df.head()
```

```python
# Display dataset information

df.info()
```

```python
# Create TF-IDF vectorizer
# max_features keeps the 5000 most important features
# ngram_range=(1,2) extracts both unigrams and bigrams

tfidf_vectorizer = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1,2)
)
```

```python
# Generate the TF-IDF feature matrix

X_tfidf = tfidf_vectorizer.fit_transform(df["Processed Text"])
```

```python
# Display the dimensions of the TF-IDF matrix

print("Number of Reviews :", X_tfidf.shape[0])
print("Number of Features:", X_tfidf.shape[1])
```

```python
# Get all extracted feature names

feature_names = tfidf_vectorizer.get_feature_names_out()

# Display the first 30 extracted features

print(feature_names[:50])
```

```python
# Display only alphabetic features for better readability

alphabetic_features = [
    feature for feature in feature_names
    if feature[0].isalpha()
]

print(alphabetic_features[:30])
```

```python
# Extract unigram features (single words)

unigrams = [
    feature for feature in feature_names
    if len(feature.split()) == 1
]

print("Total Unigrams:", len(unigrams))

print(unigrams[:20])
```

```python
# Extract bigram features (two-word phrases)

bigrams = [
    feature for feature in feature_names
    if len(feature.split()) == 2
]

print("Total Bigrams:", len(bigrams))

print(bigrams[:20])
```

```python
# Feature engineering completed successfully

print("TF-IDF feature matrix created successfully.")
print("Matrix Shape:", X_tfidf.shape)
```
