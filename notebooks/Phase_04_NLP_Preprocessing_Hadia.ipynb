## Task 1 (Lowercase Conversion)

```python
# Import required libraries
import pandas as pd
import re
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
# Download required NLTK packages (Run only once)
```

```python
# Download required NLTK packages (Run only once)

import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# Load the cleaned dataset

df = pd.read_csv("Cleaned_Reviews_File.csv")
```

```python
# Display the first five records

df.head()
```

```python
# Display sample reviews before preprocessing

df["Review Text"].head()
```

```python
# Convert all review text to lowercase

df["Review Text"] = df["Review Text"].str.lower()
```

```python
# Display reviews after lowercase conversion

df["Review Text"].head(10)
```

```python
# Display the first five records

df.head()
```

## Task 2 (Regex Cleaning)

```python
# Remove URLs and extra white spaces using regex

df["Review Text"] = (
    df["Review Text"]
    .str.replace(r"http\S+|www\S+", "", regex=True)   # Remove URLs
    .str.replace(r"\s+", " ", regex=True)             # Replace multiple spaces with a single space
    .str.strip()                                      # Remove leading/trailing spaces
)
```

```python
# Display sample reviews after regex cleaning

df["Review Text"].head(10)
```

## Task 3 (Remove Special Characters)

```python
# Remove special characters from review text

df["Review Text"] = df["Review Text"].str.replace(r"[^a-zA-Z0-9\s]", "", regex=True)
```

```python
# Display sample reviews after removing special characters

df["Review Text"].head(10)
```

## Task4 (Tokenization)

```python
# Tokenize review text into individual words

df["Review Tokens"] = df["Review Text"].apply(word_tokenize)
```

```python
# Display tokenized reviews

df[["Review Text", "Review Tokens"]].head()
```

## Task 5 (Stop-word Removal)

```python
# Initialize English stop words

stop_words = set(stopwords.words('english'))
```

```python
# Remove stop words from tokenized reviews

df["Review Tokens"] = df["Review Tokens"].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)
```

```python
# Display reviews after stop-word removal

df[["Review Text", "Review Tokens"]].head()
```

## Task 6 (Apply Lemmatization)

```python
# Initialize the WordNet lemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
# Apply lemmatization to tokenized reviews

df["Review Tokens"] = df["Review Tokens"].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)
```

```python
df[["Review Text", "Review Tokens"]].head(10)
```

```python
# Convert token lists back into text for TF-IDF

df["Processed Text"] = df["Review Tokens"].apply(lambda tokens: " ".join(tokens))
```

```python
# Display processed text

df[["Review Tokens", "Processed Text"]].head()
```

```python
# Save the preprocessed dataset

df.to_csv("Preprocessed_Reviews_File.csv", index=False)

print("Preprocessed dataset saved successfully.")
```
